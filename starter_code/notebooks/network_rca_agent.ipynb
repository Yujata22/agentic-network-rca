```python
# Put the starter_code directory (which contains db.py and the agent/ package)
# on sys.path, regardless of the kernel's working directory. We walk up from
# the current directory until we find db.py, so this works whether the kernel
# starts in notebooks/, in work/, or elsewhere.
import sys
from pathlib import Path

start_dir = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    if (candidate / 'db.py').exists():
        start_dir = candidate
        break
if start_dir is None:
    raise RuntimeError(
        "Could not locate db.py by walking up from "
        f"{Path.cwd()}. Run this notebook from inside the mounted work/ tree."
    )
if str(start_dir) not in sys.path:
    sys.path.insert(0, str(start_dir))

from db import run_query, list_tables

print('Using starter_code dir:', start_dir)
print('Tables:', list_tables())
```

```python
for t in list_tables():
    n = run_query(f'SELECT COUNT(*) AS n FROM {t}')[0]['n']
    print(f'{t:24s} {n}')
```
